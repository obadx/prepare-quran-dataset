# Muaalem FastConformer — live microphone streaming (Colab)

True cache-aware streaming inference for `obadx/muaalem-fastconformer-base-v1` from your **browser microphone**, via a Gradio app.

Unlike the file-based `infer_fastconformer_streaming` (which mel-transforms the whole file up front), this computes mel frames **incrementally** from the live audio buffer in a way that is bit-identical to the whole-file mel, and threads the encoder cache chunk by chunk (~0.5 s of new audio per step, ~0.5 s model lookahead).

**How to use:**
1. Runtime → Change runtime type → **GPU** (CPU works too, just slower).
2. Run all cells. The self-test cell verifies the streaming path against the trusted file simulation — expect `SELF TEST PASSED`.
3. Open the Gradio **share link**, allow microphone access, and speak. Stop recording to flush the final chunk.

**Diagnostics.** The app once froze mid-session with
`RuntimeError: Input after last input` from `soxr`; that is fixed in Part C below, and it
was the bug. The three parts now serve as a standing harness: **A** proves the chunked-mel
algebra, **B** guards the settled gain/resampling results and measures what dropped audio
costs, and **C** runs an instrumented live session and checks the live predictions against
a replay of the same buffer.

# Part A — the chunked-mel algebra

`tests/test_streaming_processor.py` derives, from scratch, how to cut a waveform into
chunks so that a per-chunk mel spectrogram equals the whole-file one. It is an
*independent* derivation from the one the mic app uses, which makes it a cross-check
rather than a restatement.

With this repo's mel config (`hop=160`, `n_fft=512`, `window=400`, `front_pad=120`):

| quantity | formula | value |
|---|---|---|
| `overlap` | `window - hop` | **240** samples |
| `drop_start` | `(front_pad + window//2) // hop` | **2** frames |
| `drop_end` | `window // (2*hop)` | **1** frame |

**Why `drop_start = 2`.** The 400-sample Hann window sits inside a 512-point FFT, so
frame `f`'s *weighted* span is `[f*160 - 200, f*160 + 200)`. Real audio starts at index
`front_pad = 120`, so a frame is uncontaminated iff `f*160 - 200 >= 120`, i.e. `f >= 2`.
`front_pad = 120` is not arbitrary — it is chosen to make `front_pad + window//2 = 320`
an exact multiple of `hop`, which is what `check_front_pad` asserts.

**Why this matters for the mic app.** `StreamingMelFeaturizer` uses a different strategy:
`front_pad = 0`, and it slices `MARGIN_FRAMES` frames off the front of a deliberately
wide window. The same geometric argument (plus NeMo's preemphasis first-sample special
case at window index 0) bounds contamination at **2 frames** — so `MARGIN_FRAMES = 8` is
a proven 4x safety margin, not a guess. That is *why* the `self_test` mel-exactness
check passes at `1e-5`.

**The debugging conclusion:** the feature pipeline is exact by construction and verified
numerically below. A blank transcript is therefore **not** a mel bug — which is what
sends us to Part B.

````python
# Part A: run the chunked-mel algebra against the model's OWN processor kwargs
# (a strictly stronger test than the module's __main__, which uses bare defaults).
import inspect

import torch
from test_streaming_processor import (
    StreamingProcessorMath,
    run_offline_processor,
    run_streaming_processor,
)

from prepare_quran_dataset.modeling_fastconformer_cache_aware import (
    FastConformerMelProcessor,
)
from streaming_mic_fastconformer import MARGIN_FRAMES

print(inspect.getsource(StreamingProcessorMath))

sp = StreamingProcessorMath()
proc_math = FastConformerMelProcessor(
    **{**model.config.processor_kwargs, "dither": 0.0, "pad_to": 0}
)

frames_per_chunk = 4  # 4 mel frames = 1 encoder output frame at 4x subsampling
chunk = sp.calc_chunk_from_frames(frames_per_chunk)

# The two invariants calc_chunk_from_frames is supposed to satisfy.
net = sp.calc_samples_to_frames(sp.front_pad + chunk) - sp.drop_start - sp.drop_end
assert net == frames_per_chunk, (
    f"calc_chunk_from_frames nets {net} frames, expected {frames_per_chunk}"
)
assert chunk - sp.overlap == sp.hop * frames_per_chunk, (
    f"chunk stride {chunk - sp.overlap} != hop * frames {sp.hop * frames_per_chunk}"
)

print(f"overlap      : {sp.overlap} samples")
print(f"drop_start   : {sp.drop_start} frames")
print(f"drop_end     : {sp.drop_end} frames")
print(f"chunk        : {chunk} samples ({chunk / 16:.1f} ms) -> {frames_per_chunk} frames")

torch.manual_seed(0)
wave = torch.randn(1, 320000)  # 20 s
offline_feats, offline_windows = run_offline_processor(sp, wave, proc_math)
stream_feats, stream_windows = run_streaming_processor(sp, wave, proc_math, chunk=chunk)

margin = min(offline_feats.shape[2], stream_feats.shape[2])
off, stream = offline_feats[:, :, :margin], stream_feats[:, :, :margin]
print(f"offline mel  : {tuple(offline_feats.shape)}")
print(f"streaming mel: {tuple(stream_feats.shape)}")
print(f"max |diff|   : {(off - stream).abs().max().item():.3e}")
print(f"allclose(atol=1e-6): {torch.allclose(off, stream, atol=1e-6)}")
print(f"allclose(default)  : {torch.allclose(off, stream)}")

# The tie-back to the mic app: the algebra bounds contamination at drop_start frames.
assert MARGIN_FRAMES >= sp.drop_start
print(
    f"\nmic app MARGIN_FRAMES = {MARGIN_FRAMES} vs proven contamination "
    f"{sp.drop_start} frames ({MARGIN_FRAMES // sp.drop_start}x margin)"
)
````

# Part B — regression harness

This started as an investigation and is now a regression guard. Two hypotheses are
**settled by measurement**, and are kept below only as cheap spot checks:

- **Level (`normalize="NA"`) — not the problem.** The non-blank frame ratio is flat
  (0.143–0.167) across a 48 dB sweep and the transcript is character-identical at every
  gain. The mean log-mel does move by exactly the predicted `0.2303 nats/dB`, confirming
  the mechanism is real — the model simply tolerates it. Training audio sits at
  −6 to −16 dBFS and a browser mic at ≈ −22 dBFS, which turns out to be well inside
  tolerance.
- **Resampling — not the problem.** A 48 kHz round-trip through the app's own stateful
  `soxr.ResampleStream` drifts by 0 samples and leaves the transcript unchanged.

**One question stays open:** a live session captured 30.0 s of audio over 37.3 s of wall
clock. Does losing audio like that cost accuracy? B2 answers it with a sweep.

> B2 previously reported "not reproduced" **having tested nothing**: it excised 100 ms
> every 2 s from a buffer that `librosa` loaded as 1.6 s, so the excision loop never ran.
> It now asserts that it actually removed something, and runs on `test.wav` (8.8 s, and a
> WAV — `test_sample.mp3` loads as 1.6 s of a 3.9 s file, which also quietly shortens
> `self_test`). The duration table below makes that truncation visible.

````python
# Part B: regression harness. Gain and resampling are settled (see the notes above);
# what remains open is whether dropped audio costs accuracy.
import difflib
import subprocess

import librosa
import numpy as np
import soxr
import torch

from prepare_quran_dataset.modeling_fastconformer_cache_aware import (
    FastConformerMelProcessor,
    infer_fastconformer_streaming,
)
from streaming_mic_fastconformer import (
    BLANK_ID,
    SAMPLE_RATE,
    FastConformerMicStreamer,
    ctc_collapse,
)

ASSETS = "/content/prepare-quran-dataset/assets/audio-sampels/"
REF_PATH = ASSETS + "test_sample.mp3"  # the file self_test gates on
GAP_PATH = ASSETS + "test.wav"  # longest asset, and a WAV: no mp3 decoder ambiguity

PH = "phonemes" if "phonemes" in vocab else next(iter(vocab))
ID_TO_TOK = {i: t for t, i in vocab[PH].items()}
ref_proc = FastConformerMelProcessor(
    **{**model.config.processor_kwargs, "dither": 0.0, "pad_to": 0}
).to(device)

def probe_duration(path: str) -> float | None:
    """Independent decode reference.

    soundfile and librosa both sit on libsndfile, so comparing those two cannot catch a
    short mp3 decode: for test_sample.mp3 they agree on 1.64 s while ffmpeg decodes
    3.92 s from the same bytes. ffprobe is the outside opinion.
    """
    try:
        return float(
            subprocess.run(
                ["ffprobe", "-v", "error", "-show_entries", "format=duration",
                 "-of", "csv=p=0", path],
                capture_output=True, text=True, timeout=30, check=True,
            ).stdout.strip()
        )
    except Exception:
        return None


print(f"{'asset':<20} {'librosa':>9} {'ffprobe':>9} {'rms dBFS':>9}")
for path in (REF_PATH, GAP_PATH, ASSETS + "test_sample_3.mp3"):
    y, _ = librosa.load(path, sr=SAMPLE_RATE, mono=True)
    loaded = len(y) / SAMPLE_RATE
    real = probe_duration(path)
    shown = "      n/a" if real is None else f"{real:8.2f}s"
    flag = ""
    if real is not None and abs(loaded - real) > 0.2:
        flag = f"  <-- librosa reads {loaded / real:.0%} of it"
    print(f"{path.split('/')[-1]:<20} {loaded:>8.2f}s {shown}"
          f" {20 * np.log10(np.sqrt(np.mean(y**2))):>+8.1f}{flag}")
print("self_test loads test_sample.mp3 through librosa.load, so it gates on whatever "
      "that returns.")

ref_wav, _ = librosa.load(REF_PATH, sr=SAMPLE_RATE, mono=True)
gap_wav, _ = librosa.load(GAP_PATH, sr=SAMPLE_RATE, mono=True)


def dbfs(x: np.ndarray) -> tuple[float, float]:
    """(rms, peak) in dBFS, floored so silence does not blow up the log."""
    x = np.asarray(x, dtype=np.float32)
    to_db = lambda v: 20.0 * np.log10(v) if v > 1e-12 else -np.inf
    return (
        to_db(float(np.sqrt(np.mean(x**2))) if x.size else 0.0),
        to_db(float(np.max(np.abs(x))) if x.size else 0.0),
    )


def tokens(ids: list[int]) -> str:
    return "".join(ID_TO_TOK.get(t, "?") for t in ctc_collapse(ids))


def nonblank(ids: list[int]) -> float:
    return sum(1 for i in ids if i != BLANK_ID) / max(len(ids), 1)


@torch.no_grad()
def offline_probe(wav: np.ndarray) -> tuple[str, float, float]:
    """Whole-file forward. Returns (transcript, non-blank ratio, mean log-mel)."""
    t = torch.tensor([wav], dtype=torch.float32, device=device)
    n = torch.tensor([len(wav)], dtype=torch.long, device=device)
    ids = model(raw_audio=t, audio_length=n).logits[PH][0].argmax(dim=-1).tolist()
    mel, _ = ref_proc(input_signal=t, length=n)
    return tokens(ids), nonblank(ids), float(mel.mean())


def stream_probe(wav: np.ndarray, chunk_s: float = 0.5) -> tuple[str, float, list[int]]:
    """Same buffer through the live mic streamer. Fresh streamer: the cache must not leak."""
    st = FastConformerMicStreamer(model, vocab, device)
    step = int(chunk_s * SAMPLE_RATE)
    for pos in range(0, len(wav), step):
        st.feed(wav[pos : pos + step])
    st.flush()
    ids = st.level_ids[PH]
    return tokens(ids), nonblank(ids), ids


def similarity(a: str, b: str) -> float:
    return difflib.SequenceMatcher(None, a, b).ratio()


def make_gapped(wav: np.ndarray, period_s: float, drop_ms: float) -> tuple[np.ndarray, float]:
    """Excise `drop_ms` every `period_s` and concatenate — what a dropped browser chunk
    looks like to the model. Asserts it actually removed something: the first version of
    this test used a period longer than the file and silently tested nothing."""
    keep = np.ones(len(wav), dtype=bool)
    width = int(drop_ms / 1000 * SAMPLE_RATE)
    starts = range(int(period_s * SAMPLE_RATE), len(wav), int(period_s * SAMPLE_RATE))
    for start in starts:
        keep[start : start + width] = False
    removed = 1.0 - keep.mean()
    assert removed > 0.0, (
        f"gap test removed nothing: {len(wav) / SAMPLE_RATE:.1f}s audio, "
        f"{period_s}s period — the file is shorter than one period"
    )
    return wav[keep], removed


# --- B1: gain. Settled -- flat across 48 dB. Kept as a cheap regression guard. -----
print("\n--- B1  gain spot check (hypothesis 1: SETTLED, expect flat) ---")
_, _, base_mel = offline_probe(ref_wav)
for db in (12, 0, -24):
    text, ratio, mel_mean = offline_probe(ref_wav * (10.0 ** (db / 20.0)))
    print(f"{db:>+4}dB | non-blank {ratio:.3f} | log-mel {mel_mean:+6.2f} "
          f"(expect {base_mel + 0.2303 * db:+.2f}) | {text[:40]}")

# --- B2: does dropped audio cost accuracy? The one question still open. ------------
print(f"\n--- B2  dropped-chunk sweep on {GAP_PATH.split('/')[-1]} "
      f"({len(gap_wav) / SAMPLE_RATE:.1f}s) ---")
intact_text, intact_ratio, _ = stream_probe(gap_wav)
print(f"{'removed':>8} {'non-blank':>10} {'vs intact':>10}  transcript")
print(f"{0.0:>7.1%} {intact_ratio:>10.3f} {1.0:>10.1%}  {intact_text[:44]}")
for period_s, drop_ms in ((2.0, 100), (1.0, 100), (0.5, 100), (1.0, 500)):
    gapped, removed = make_gapped(gap_wav, period_s, drop_ms)
    text, ratio, _ = stream_probe(gapped)
    print(f"{removed:>7.1%} {ratio:>10.3f} {similarity(intact_text, text):>10.1%}"
          f"  {text[:44]}")
print("Read this as the accuracy cost of the wall-clock deficit seen in a live session.")

# --- B3: the app's own stateful chunked resampler, 48k -> 16k ----------------------
print("\n--- B3  48 kHz round-trip through soxr.ResampleStream (hypothesis 5: SETTLED) ---")
up = soxr.resample(ref_wav, SAMPLE_RATE, 48000)
rs = soxr.ResampleStream(48000, SAMPLE_RATE, 1, dtype="float32")
step = int(0.5 * 48000)
parts = [rs.resample_chunk(up[p : p + step]) for p in range(0, len(up), step)]
parts.append(rs.resample_chunk(np.zeros(0, dtype=np.float32), last=True))
round_trip = np.concatenate([p for p in parts if len(p)]).astype(np.float32)
rt_text, rt_ratio, _ = stream_probe(round_trip)
ref_text, ref_ratio, _ = stream_probe(ref_wav)
print(f"length drift {len(round_trip) - len(ref_wav):+d} samples | "
      f"non-blank {rt_ratio:.3f} vs {ref_ratio:.3f} | "
      f"transcript match {similarity(ref_text, rt_text):.1%}")
````

# Part C — the live session

**The crash below was the bug.** The previous `build_demo` ran `_resample(...)` one line
*above* its `finished` guard:

```python
y16k = _resample(sr, _to_float_mono(y))           # <-- threw here
if len(y16k) and not state["streamer"].finished:  # <-- guard never reached
```

`on_stop` finalises the `soxr.ResampleStream` with `last=True`. Gradio then delivers
stream events that were already queued, so `resample_chunk` hits a finalised stream and
soxr raises `RuntimeError: Input after last input`. From that moment every stream event
died and the UI froze — the "it stops when I pause" symptom. Below, the `finished` check
comes first and finalisation is a one-way latch. With that fix the app produces real
phonemes live.

### Recording protocol

1. **Reload the share link** so the `getUserMedia` patch (AGC / noise-suppression off) applies.
2. Recite ~30 s. Gain is settled as a non-issue, so the slider can stay at 0 dB.
3. Press the app's **stop button** — that is what flushes the final chunk and saves the capture.
4. Run the post-mortem cell.

The post-mortem's job is one comparison: the ids the **live** session emitted versus a
**replay** of the very same buffer through the path `self_test` verifies. `feed()` is
chunk-size invariant, so they must agree exactly. If they do, the streaming path is
proven end to end on real microphone audio — which `self_test`, seeing only files, can
never show.

````python
# Part C: the fixed + instrumented mic app.
#
# Defined here rather than edited into tests/streaming_mic_fastconformer.py so the
# module's self_test contract stays exactly as verified in cell 5.
import time
import traceback

import gradio as gr
import numpy as np
import soundfile as sf
import soxr

from streaming_mic_fastconformer import (
    BLANK_ID,
    EOS_ID,
    SAMPLE_RATE,
    FastConformerMicStreamer,
)

CAPTURE_PATH = "/content/mic_capture.wav"
TARGET_DBFS = -12.0  # low end of the band the training audio occupies

LAST_SESSION: dict[str, object] = {}
"""Forensics hand-off: demo.launch() runs the server in a background thread, so the
post-mortem cell can inspect a session once it has been stopped."""

# Browser WebRTC defaults (AGC / noise suppression / echo cancellation) are tuned for
# telephony and mangle recitation. Gradio exposes no getUserMedia constraints, so the
# page patches the API itself. Requires a hard reload of the share link to take effect.
MIC_CONSTRAINTS_JS = """
<script>
(function () {
  const md = navigator.mediaDevices;
  if (!md || !md.getUserMedia || md.__muaalemPatched) return;
  const orig = md.getUserMedia.bind(md);
  md.getUserMedia = function (c) {
    if (c && c.audio) {
      const a = (typeof c.audio === "object") ? c.audio : {};
      c = Object.assign({}, c, {audio: Object.assign({}, a, {
        echoCancellation: false, noiseSuppression: false, autoGainControl: false})});
    }
    return orig(c);
  };
  md.__muaalemPatched = true;
})();
</script>
"""


def collapse_carry(ids: list[int], prev: int) -> tuple[list[int], int]:
    """Greedy CTC collapse that carries `prev` across a chunk boundary.

    Restarting `prev` at every chunk — which `ctc_collapse` does, since it always
    begins at blank — re-emits a phoneme whose frames straddle the boundary. In this
    scheme that is not cosmetic: a doubled token is a geminate (shadda) and the repeat
    count is duration in harakat, so a seam duplicate invents a shadda.

    Specials are filtered *after* collapsing, matching train_streaming.ctc_decode:
    [EOS] separates two identical phonemes rather than merging them.
    """
    out: list[int] = []
    for t in ids:
        if t == BLANK_ID:
            prev = BLANK_ID
            continue
        if t != prev:
            out.append(t)
        prev = t
    return [t for t in out if t not in (BLANK_ID, EOS_ID)], prev


def build_demo_v2(model, vocab, device):
    ph = "phonemes" if "phonemes" in vocab else next(iter(vocab))
    state: dict = {}

    def _fresh(status: str):
        state.clear()
        state.update(
            streamer=FastConformerMicStreamer(model, vocab, device),
            resampler=None,
            native_sr=None,
            resampler_closed=False,
            delivered_sr=None,
            pending=[],
            arrivals=[],
            log=[],
            errors=[],
            dtypes=set(),
            t0=time.perf_counter(),
            status=status,
            capture=None,
            consumed=0,
            prev=BLANK_ID,
        )

    _fresh("idle")

    def _dbfs(x: np.ndarray) -> tuple[float, float]:
        rms = float(np.sqrt(np.mean(x**2))) if x.size else 0.0
        peak = float(np.max(np.abs(x))) if x.size else 0.0
        to_db = lambda v: 20.0 * np.log10(v) if v > 1e-12 else -np.inf
        return to_db(rms), to_db(peak)

    def _to_float_mono(y: np.ndarray) -> np.ndarray:
        if y.dtype == np.int16:
            y = y.astype(np.float32) / 32768.0
        elif y.dtype == np.int32:
            y = y.astype(np.float32) / 2147483648.0
        else:
            y = np.asarray(y, dtype=np.float32)
        return y.mean(axis=1) if y.ndim > 1 else y

    def _resample(sr: int, y: np.ndarray, last: bool = False) -> np.ndarray:
        """Stateful 16 kHz resample. Refuses a finalised stream — soxr raises
        'Input after last input', which is what killed the original app."""
        if state["resampler_closed"]:
            return np.zeros(0, dtype=np.float32)
        if sr == SAMPLE_RATE and not last:
            return y
        if state["resampler"] is None or state["native_sr"] != sr:
            state["resampler"] = soxr.ResampleStream(sr, SAMPLE_RATE, 1, dtype="float32")
            state["native_sr"] = sr
        out = state["resampler"].resample_chunk(y, last=last)
        if last:
            state["resampler_closed"] = True
        return out

    def _drain() -> None:
        """Run inference on everything that has arrived since the last tick."""
        st = state["streamer"]
        if st.finished or not state["pending"]:
            return
        pending, state["pending"] = state["pending"], []
        samples = np.concatenate(pending)
        rms, peak = _dbfs(samples)

        before = len(st.chunk_log)
        st.feed(samples)
        for entry in st.chunk_log[before:]:
            ids = st.level_ids[ph][state["consumed"] : state["consumed"] + entry["frames_out"]]
            state["consumed"] += entry["frames_out"]
            toks, state["prev"] = collapse_carry(ids, state["prev"])
            state["log"].append(
                {
                    **entry,
                    "rms": rms,
                    "peak": peak,
                    "phonemes": "".join(st.id_to_token[ph].get(t, "?") for t in toks),
                }
            )

    def _render(capture: str | None = None):
        capture = capture or state.get("capture")
        st = state["streamer"]
        ids = st.level_ids[ph]
        blank = sum(1 for i in ids if i == BLANK_ID) / max(len(ids), 1)
        fed, wall = st.seconds_fed(), time.perf_counter() - state["t0"]
        lat = [e["latency_ms"] for e in st.chunk_log]

        summary = (
            f"{state['status']} | steps {st.step} | fed {fed:.1f}s / wall {wall:.1f}s "
            f"(deficit {wall - fed:+.1f}s) | blank {blank:.1%} | "
            f"avg {np.mean(lat) if lat else 0:.0f} ms | sr {state['delivered_sr']} | "
            f"dtypes {sorted(str(d) for d in state['dtypes']) or '-'}"
            + (" | FLUSHED" if st.finished else "")
        )

        last = state["log"][-1] if state["log"] else None
        level = (
            "waiting for audio..."
            if last is None
            else f"rms {last['rms']:+.1f} dBFS | peak {last['peak']:+.1f} dBFS | "
            f"suggested gain {TARGET_DBFS - last['rms']:+.0f} dB "
            f"(training audio sits at -6 to -16 dBFS)"
        )

        log = "\n".join(
            f"step {e['step']:03d} | in {e['frames_in']:3d} mel | out {e['frames_out']:3d} "
            f"| {e['latency_ms']:6.1f} ms | {e['rms']:+6.1f} dBFS | {e['phonemes']}"
            for e in state["log"][-30:]
        )
        errors = "\n\n".join(state["errors"][-3:])
        return st.transcript(), st.sifat_summary(), log, level, summary, errors, capture

    def _guard(fn):
        """Render exceptions into the UI. Gradio buries handler tracebacks in the
        server log, which is invisible behind a share link."""
        def wrapper(*args):
            try:
                return fn(*args)
            except Exception:
                state.setdefault("errors", []).append(traceback.format_exc())
                state["status"] = "ERROR"
                return _render()
        return wrapper

    @_guard
    def on_start():
        _fresh("recording")
        return _render()

    def on_stream(audio_chunk, gain_db):
        """Ingestion only: no model call, no render.

        Gradio drops stream events that arrive while the handler is still running, so
        every millisecond spent here is potentially lost audio. Inference happens on the
        timer tick instead. Declared with outputs=[] so there is nothing to update.
        """
        try:
            st = state["streamer"]
            # Ordering matters: Gradio delivers queued stream events after stop_recording,
            # and touching the finalised resampler raised 'Input after last input'.
            if st.finished or audio_chunk is None:
                return
            sr, y = audio_chunk
            state["delivered_sr"] = sr
            state["dtypes"].add(y.dtype)
            y16k = _resample(sr, _to_float_mono(y))
            if not len(y16k):
                return
            y16k = y16k * (10.0 ** (gain_db / 20.0))
            state["pending"].append(y16k)
            # Every arrival, not just the ones that complete a mel chunk -- otherwise
            # the timeline cannot account for the audio we are trying to find.
            state["arrivals"].append((time.perf_counter() - state["t0"], len(y16k)))
        except Exception:
            state.setdefault("errors", []).append(traceback.format_exc())

    @_guard
    def on_tick():
        _drain()
        return _render()

    @_guard
    def on_stop():
        st = state["streamer"]
        if st.finished:
            return _render()
        if state["resampler"] is not None and not state["resampler_closed"]:
            tail = _resample(state["native_sr"], np.zeros(0, dtype=np.float32), last=True)
            if len(tail):
                state["pending"].append(tail)
                state["arrivals"].append((time.perf_counter() - state["t0"], len(tail)))
        _drain()
        st.flush()
        state["status"] = "stopped"

        wav = np.array(st.featurizer.buf, copy=True)
        if len(wav):
            sf.write(CAPTURE_PATH, wav, SAMPLE_RATE)
        state["capture"] = CAPTURE_PATH if len(wav) else None
        LAST_SESSION.update(
            wav16k=wav,
            path=CAPTURE_PATH,
            chunk_log=list(state["log"]),
            arrivals=list(state["arrivals"]),
            wall_seconds=time.perf_counter() - state["t0"],
            native_sr=state["delivered_sr"],
            dtypes=set(state["dtypes"]),
            errors=list(state["errors"]),
            level_ids={k: list(v) for k, v in st.level_ids.items()},
        )
        return _render()

    @_guard
    def on_reset():
        _fresh("reset")
        return _render()

    with gr.Blocks(title="Muaalem FastConformer live streaming", head=MIC_CONSTRAINTS_JS) as demo:
        gr.Markdown(
            "## Muaalem FastConformer — live cache-aware streaming (instrumented)\n"
            "Audio ingestion and inference run on separate handlers, so a slow step "
            "cannot cost you microphone audio. Press the **stop** button when done — "
            "that is what flushes the final chunk and saves the capture."
        )
        audio_input = gr.Audio(
            sources=["microphone"], streaming=True, type="numpy", label="Microphone"
        )
        gain = gr.Slider(-6, 36, value=0, step=1, label="Input gain (dB)")
        level_box = gr.Textbox(label="Input level", lines=1)
        summary_box = gr.Textbox(label="Status", lines=2)
        transcript_box = gr.Textbox(label="Phonetic transcript", lines=4, rtl=True)
        sifat_box = gr.Textbox(label="Sifat (latest)", lines=8, rtl=True)
        log_box = gr.Textbox(label="Per-chunk log (last 30)", lines=14)
        error_box = gr.Textbox(label="Errors", lines=8)
        capture_box = gr.Audio(label="Captured 16 kHz buffer", type="filepath", interactive=False)
        reset_btn = gr.Button("Reset")

        outputs = [transcript_box, sifat_box, log_box, level_box, summary_box,
                   error_box, capture_box]
        audio_input.start_recording(fn=on_start, outputs=outputs)
        # outputs=[] keeps the ingestion path free of any UI work.
        audio_input.stream(fn=on_stream, inputs=[audio_input, gain], outputs=[],
                           stream_every=0.5)
        audio_input.stop_recording(fn=on_stop, outputs=outputs)
        reset_btn.click(fn=on_reset, outputs=outputs)
        # If gr.Timer is missing in this Gradio build, fall back to calling _drain()
        # from on_stream and wiring outputs back onto .stream().
        gr.Timer(0.5).tick(fn=on_tick, outputs=outputs)

    return demo
````

````python
# Post-mortem on the live session. Requires Part B's helpers, and a session that was
# ended with the app's stop button.
import IPython.display as ipd

assert LAST_SESSION, "no session captured — record, then press the app's stop button"

errors = LAST_SESSION["errors"]
print(f"=== errors: {len(errors)} ===")
if errors:
    print(errors[-1])

wav = np.asarray(LAST_SESSION["wav16k"], dtype=np.float32)
rms, peak = dbfs(wav)
wall = float(LAST_SESSION["wall_seconds"])
captured = len(wav) / SAMPLE_RATE
deficit = wall - captured
print(f"\ncaptured : {captured:.1f}s over {wall:.1f}s wall clock (deficit {deficit:+.1f}s)")
print(f"level    : rms {rms:+.1f} dBFS, peak {peak:+.1f} dBFS "
      f"(reference file {dbfs(ref_wav)[0]:+.1f} dBFS)")
print(f"browser  : sr {LAST_SESSION['native_sr']}, dtypes "
      f"{sorted(str(d) for d in LAST_SESSION['dtypes'])}")

# --- where the deficit went ------------------------------------------------------
# Each arrival (t, n) means n samples landed at t. The chunk therefore covers roughly
# [t - n/sr, t], so the three ways to lose audio are: before the first chunk's span
# (warm-up), between consecutive spans (drops), and after the last one (stop latency).
arrivals = LAST_SESSION.get("arrivals") or []
print("\n--- deficit attribution ---")
if not arrivals:
    print("  no arrivals recorded (re-run the app cell: this needs the new instrumentation)")
else:
    ts = np.array([t for t, _ in arrivals])
    ns = np.array([n for _, n in arrivals]) / SAMPLE_RATE
    warmup = max(0.0, ts[0] - ns[0])
    tail = max(0.0, wall - ts[-1])
    span = ts[-1] - (ts[0] - ns[0])
    drops = max(0.0, span - ns.sum())
    inter = ts[1:] - ts[:-1] - ns[1:]  # audio missing between consecutive chunks
    print(f"  chunks received   : {len(arrivals)} ({ns.sum():.1f}s of audio)")
    print(f"  warm-up           : {warmup:5.1f}s  (start_recording -> first audio)")
    print(f"  drops mid-stream  : {drops:5.1f}s  (gaps between chunks)")
    print(f"  stop latency      : {tail:5.1f}s  (last chunk -> stop handler)")
    print(f"  accounted         : {warmup + drops + tail:5.1f}s of {deficit:.1f}s deficit")
    if len(inter):
        worst = int(np.argmax(inter))
        print(f"  largest single gap: {inter[worst]:5.1f}s at t={ts[worst + 1]:.1f}s")
    biggest = max((warmup, "warm-up"), (drops, "mid-stream drops"), (tail, "stop latency"))
    print(f"  -> dominated by {biggest[1]} ({biggest[0]:.1f}s)")
    if biggest[1] != "mid-stream drops":
        print("     Decoupling ingestion from inference does NOT recover these seconds;")
        print("     the remedy is a lead-in delay before reciting / stopping later.")

ipd.display(ipd.Audio(wav, rate=SAMPLE_RATE))

# --- live vs replay: the decisive correctness check -------------------------------
live_ids = list(LAST_SESSION["level_ids"][PH])
replay_text, replay_ratio, replay_ids = stream_probe(wav)
live_text, live_ratio = tokens(live_ids), nonblank(live_ids)

same_len = len(live_ids) == len(replay_ids)
n = min(len(live_ids), len(replay_ids))
first_diff = next((i for i in range(n) if live_ids[i] != replay_ids[i]), None)
frame_match = sum(1 for a, b in zip(live_ids, replay_ids) if a == b) / max(n, 1)

print(f"\nlive   : {len(live_ids):>5} frames | non-blank {live_ratio:.3f}\n  {live_text[:90]}")
print(f"replay : {len(replay_ids):>5} frames | non-blank {replay_ratio:.3f}\n  {replay_text[:90]}")
print(f"frame agreement {frame_match:.4f}"
      + ("" if first_diff is None else f" | first difference at frame {first_diff}"))

# The per-chunk log must now concatenate to the transcript. It did not before: each
# chunk was collapsed from a fresh blank, so a phoneme straddling a boundary was
# emitted twice -- which in this scheme fabricates a shadda.
joined = "".join(e["phonemes"] for e in LAST_SESSION["chunk_log"])
print(f"\nper-chunk log concatenates to transcript: {joined == live_text}")
if joined != live_text:
    print(f"  log        : {joined[:90]}")
    print(f"  transcript : {live_text[:90]}")

offline_text, offline_ratio, _ = offline_probe(wav)
_, ref_ratio_off, _ = offline_probe(ref_wav)
print(f"\noffline on capture   : non-blank {offline_ratio:.3f} (reference {ref_ratio_off:.3f})")
print(f"  {offline_text[:90]}")

print("\n================ VERDICT ================")
if same_len and first_diff is None:
    print("LIVE MATCHES REPLAY — the streaming path is verified end to end on real")
    print(f"  microphone audio ({len(live_ids)} frames, exact).")
elif frame_match > 0.99:
    print(f"NEAR MATCH — {frame_match:.2%} of frames agree; lengths "
          f"{len(live_ids)} vs {len(replay_ids)}. Most likely the final partial chunk.")
else:
    print(f"LIVE DIFFERS FROM REPLAY — only {frame_match:.2%} of frames agree"
          + (f", first at frame {first_diff}" if first_diff is not None else "") + ".")
    print("  The same buffer replays correctly, so this is a live-only bug.")

if deficit > 2.0:
    print(f"\nDeficit still {deficit:+.1f}s — see the attribution above for where it goes.")
elif arrivals:
    print(f"\nDeficit {deficit:+.1f}s: the microphone path is keeping up.")
````
